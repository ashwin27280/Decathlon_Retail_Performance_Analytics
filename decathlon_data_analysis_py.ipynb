**Feature Engineering**

```python
df.columns
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
#Create Order Month
df["Order_Month"] = df["Order_Date"].dt.month_name()
```

```python
#Create Order Year
df["Order_Year"] = df["Order_Date"].dt.year
```

```python
#Create Quarter
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
#Create Day Name
df["Day_Name"] = df["Order_Date"].dt.day_name()
```

```python
#Create Profit Margin %
df["Profit_Margin"] = (df["Profit"] /df["Final_Amount"]) * 100
```

for Profitability Analysis

```python
#Create Order Value Category

df["Order_Value"] = pd.cut(
    df["Final_Amount"],
    bins=[0,1000,5000,10000,100000],
    labels=[
        "Low",
        "Medium",
        "High",
        "Premium"
    ]
)
```

for Customer Segmentation

```python
#Create Age Group

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0,18,25,35,45,60,100],
    labels=[
        "<18",
        "18-25",
        "26-35",
        "36-45",
        "46-60",
        "60+"
    ]
)
```

```python
#Create High Value Customer

threshold = df["Final_Amount"].quantile(0.75)

df["High_Value_Customer"] = np.where(
    df["Final_Amount"] >= threshold,
    "Yes",
    "No"
)
```

```python
#Now Save Dataset
df.to_csv("Decathlon_Feature_Engineered.csv",index=False)
```
